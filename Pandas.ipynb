```python
fruit_protein = {
    'Avocado'     : 2.0, #grams of protein
    'Guava'       : 2.6,
    'blackberries': 2.0,
    'oranges'     : 0.9,
    'banana'      : 1.1,
    'apples'      : 0.3,
    'kiwi'        : 1.1,
    'pomegranate' : 1.7,
    'mangoes'     : 0.8,
    'cherries'    : 1.0
}
```

```python
s2 = pd.Series(fruit_protein,name = "protein")
s2
```

```python
#conditional selecton
s2>1
```

```python
s2[s2>1]
```

```python
#logical operators: and,or,not
s2[(s2>0.5) & (s2<2)]
```

```python
s2[(s2>0.5) | (s2<2)]
```

```python
#not operation
s2[~(s2>1)]
```

```python
#modifying the series
s2['mangoe'] = 2.3
```

# data **cleaning** *italicized text*

```python
# data cleaning

df.isnull().sum()
```

```python
df.dropna(how = 'any') #or df.dropna()
```

```python
df
```

```python
df.dropna(how = 'all') #any row that had any null value
```

```python
df
```

```python
df.fillna(0)
```

```python
df['Age'].fillna(df["Age"].mean())
```

```python
df["Salary"].fillna(df["Salary"].median())
```

```python
df["Age"].fillna(method = "ffill")#forward fill
```

```python
df["Age"].fillna(method = "bfill") #backward fill
```

```python
df["Name"].replace("Charlie","Rose")
```

```python
df["Name"] = df["Name"].replace("Charlie","Rose")
```

```python
df
```

```python
#duplicates

df_dup = df[df.duplicated(keep = "first")]#(first give the last doop value) #out here will only return second doop value
df_dup
```

```python
df_dup = df[df.duplicated(keep = "last")] #(last give the first doop value)
df_dup
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
#invalid values:
#lambda -> python

df["Promoted Salary"] = df["Promoted Salary"].apply(lambda x: x/10 if x>65000 else x)
```

```python
df
```

```python
name = 'alice_fernandes'
df[['frist_name','last_name']] = df['Name'].str.split("_")
```

```python
#joins and merges
#joins -> leftjoin,rightjoin,outerjoin,innerjoin

department_info = {
    "Dept" : ["HR","IT","Finance"]                 ,
    "Location"   : ["New York","San Francisco","Chicago"],
    "Manager"    : ["Laura","Steve","Nina"]
}

df2 = pd.DataFrame(department_info)
df2
```

```python
#Concat
pd.concat([df, df2]) # -> merges two dataframes
```

```python
pd.concat([df, df2] , axis = 1)
```
